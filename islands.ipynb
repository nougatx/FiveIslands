```python
import pandas as pd
import folium

data = [['Tristan da Cunha', -37.10880642711254,-12.288562569323755 ],
                                    ['Saint Helena', -15.960240609448975, -5.7047841487715125],
                                    ['Ascension Island', -7.9423559696089026, -14.366533397347828],
                                    ['Pitcairn Islands, Adamstown', -25.06992321107777, -130.10593367045323],
                                    ['Diomede Islands', 65.78507906407131, -168.9783413436392]]

world_map = pd.DataFrame(data, columns = ['island', 'lat', 'lng'])   

map_location = [28.94904556915751, -59.072195491921946]
map = folium.Map(location=map_location, zoom_start=2)

# Iterate over the DataFrame to create markers for each island
for _, row in world_map.iterrows():
    # Create marker for each island
    location = [row['lat'], row['lng']]
    popup = f"<strong>{row['island']}</strong>"
    
    # Add marker to the map
    marker = folium.Marker(location=location, popup=popup)
    marker.add_to(map)


display(map)

```

```python
map.save('world_map.html')
```

```python
import pandas as pd
import folium

# Data for the islands
data = [['Tristan da Cunha', -37.10880642711254, -12.288562569323755],
        ['Saint Helena', -15.960240609448975, -5.7047841487715125],
        ['Ascension Island', -7.9423559696089026, -14.366533397347828],
        ['Pitcairn Islands, Adamstown', -25.06992321107777, -130.10593367045323],
        ['Diomede Islands', 65.78507906407131, -168.9783413436392]]

# Create a DataFrame with island names and coordinates
world_map = pd.DataFrame(data, columns=['island', 'lat', 'lng'])

# Set the initial map location and zoom level
map_location = [28.94904556915751, -59.072195491921946]  # Coordinates roughly in the middle of the islands
map = folium.Map(location=map_location, zoom_start=2)

# Iterate over the DataFrame to create markers for each island
for _, row in world_map.iterrows():
    location = [row['lat'], row['lng']]
    tooltip = row['island']  # Tooltip to show island names on hover
    
    # Add marker with tooltip to the map
    marker = folium.Marker(location=location, tooltip=tooltip)
    marker.add_to(map)

# Display the map with tooltips
display(map)
```

```python
import pandas as pd
import folium

# Data for the islands
data = [['Tristan da Cunha', -37.10880642711254, -12.288562569323755],
        ['Saint Helena', -15.960240609448975, -5.7047841487715125],
        ['Ascension Island', -7.9423559696089026, -14.366533397347828],
        ['Pitcairn Islands, Adamstown', -25.06992321107777, -130.10593367045323],
        ['Diomede Islands', 65.78507906407131, -168.9783413436392]]

# Create a DataFrame with island names and coordinates
world_map = pd.DataFrame(data, columns=['island', 'lat', 'lng'])

# Set the initial map location and zoom level
map_location = [28.94904556915751, -59.072195491921946]  # Coordinates roughly in the middle of the islands
map = folium.Map(location=map_location, zoom_start=2)

# Iterate over the DataFrame to create markers for each island
for _, row in world_map.iterrows():
    location = [row['lat'], row['lng']]
    
    # Add marker to the map
    marker = folium.Marker(location=location)
    marker.add_to(map)
    
    # Add island name as a text label on the map using DivIcon
    folium.map.Marker(
        location=location,
        icon=folium.DivIcon(html=f"""<div style="font-size: 12px; color: black;"><b>{row['island']}</b></div>""")
    ).add_to(map)

# Display the map with markers and labels
display(map)
```

```python
map.save('islands.html')
```
